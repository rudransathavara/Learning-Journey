# Introduction to Microsoft Excel
### My Excel Learning Journey - Lesson 01

## 1. What Is Microsoft Excel?

Microsoft Excel is a spreadsheet application used to organise, calculate, analyse, and visualise data.


## 2. Worksheets and Workbooks

| Term | Meaning |
|---|---|
| Workbook | An Excel file containing one or more worksheets |
| Worksheet | A spreadsheet within a workbook |


## 3. Cells

| Term | Meaning |
|---|---|
| Row | A horizontal line of cells |
| Column | A vertical line of cells |
| Cell | A single box where data is entered |
| Cell Reference | The address of a cell, such as A1 |

A cell in column B and row 3 is called **B3**.
